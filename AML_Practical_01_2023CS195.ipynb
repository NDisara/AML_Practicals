```python
# Question 10: Conditional Expressions
x = 15
y = 20

if x > 10 and y < 25:
    print("Condition 1")

if x > 20 or y == 20:
    print("Condition 2")

if not (x == 15):
    print("Condition 3")
else:
    print("Condition 4")
```

Explanation:
- Condition 1: x > 10 is true and y < 25 is true. Therefore, True and True is true.
- Condition 2: x > 20 is false, but y == 20 is true. Therefore, False or True is true.
- Condition 3: x == 15 is true, so not True becomes false. The statement does not execute.
- Condition 4: The else block executes because the condition for Condition 3 is false.

```python
# Question 11: For Loop
numbers = [5, 10, 15, 20, 25]

# Print every element
for number in numbers:
    print(number)

# Calculate the sum without using sum()
total = 0

for number in numbers:
    total += number

print("Total:", total)

# Count elements greater than 12
count = 0

for number in numbers:
    if number > 12:
        count += 1

print("Count greater than 12:", count)
```

```python
# Question 12: While Loop

number = 1
total = 0

while number <= 10:
    print(number)
    total += number
    number += 1

print("Sum:", total)
```

```python
# Question 13: Filtering Values Using Loops

marks = [45, 78, 32, 90, 56, 28, 65]

passing_marks = []
failures = 0

for mark in marks:
    if mark >= 50:
        passing_marks.append(mark)
    else:
        failures += 1

average = sum(passing_marks) / len(passing_marks)

print("Passing Marks:", passing_marks)
print("Number of Failures:", failures)
print("Average Passing Marks:", average)
```

```python
# Question 14: Simple Function

def calculate_area(length, width):
    area = length * width
    return area

result = calculate_area(10, 5)

print("Area:", result)
```

```python
# Question 15: Calculating the Mean

def calculate_mean(numbers):
    total = 0

    for number in numbers:
        total += number

    mean = total / len(numbers)
    return mean

numbers = [10, 20, 30, 40, 50]

print("Mean:", calculate_mean(numbers))
```

```python
# Question 16: Functions and Conditional Statements

def check_pass(marks):
    if marks >= 50:
        return "Pass"
    else:
        return "Fail"

marks = [75, 40, 65, 30, 90]

for mark in marks:
    print(mark, "-", check_pass(mark))
```

```python
# Question 20: Student Performance Analysis

import pandas as pd
import matplotlib.pyplot as plt

# Given student dataset
students = [
    {"name": "Amal", "marks": 75},
    {"name": "Nimal", "marks": 45},
    {"name": "Kamal", "marks": 85},
    {"name": "Sunil", "marks": 35},
    {"name": "Saman", "marks": 90}
]

# 1. Extract all marks into a list
marks = []

for student in students:
    marks.append(student["marks"])

print("All Marks:", marks)

# 2. Calculate average marks
average_marks = sum(marks) / len(marks)
print("Average Marks:", average_marks)

# 3. Find highest and lowest marks
print("Highest Marks:", max(marks))
print("Lowest Marks:", min(marks))

# 4. Identify students who passed
print("Students Who Passed:")

for student in students:
    if student["marks"] >= 50:
        print(student["name"], "-", student["marks"])

# 5. Count the number of students who failed
failed_count = 0

for student in students:
    if student["marks"] < 50:
        failed_count += 1

print("Number of Failures:", failed_count)

# 6. Create a Pandas DataFrame
df = pd.DataFrame(students)

print("\nStudent DataFrame:")
print(df)

# 7. Display a bar chart showing student marks
plt.bar(df["name"], df["marks"])

plt.title("Student Performance Analysis")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.ylim(0, 100)

plt.show()
```
